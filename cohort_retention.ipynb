```python
import pandas as pd
import numpy as np
```

```python

# Read CSV from SQL export
df = pd.read_csv('E:\\Churn analysis\\Cohort_analysis.csv')

print("✅ Loaded cohort data:", df.shape[0], "rows")

# Aggregate by tenure_cohort only (ignore contract/payment for now)
cohort_summary = df.groupby('tenure_cohort').agg({
    'total_customers': 'sum',
    'churned_customers': 'sum',
    'retained_customers': 'sum'
}).reset_index()

# Calculate overall churn/retention for each cohort
cohort_summary['overall_churn_rate'] = (cohort_summary['churned_customers'] / cohort_summary['total_customers'] * 100).round(2)
cohort_summary['overall_retention_rate'] = (100 - cohort_summary['overall_churn_rate']).round(2)

# Reorder cohorts
cohort_order = ['0_months', '1_to_3_months', '4_to_6_months', '7_to_12_months', '13_to_24_months', '25plus_months']
cohort_summary['tenure_cohort'] = pd.Categorical(cohort_summary['tenure_cohort'], categories=cohort_order, ordered=True)
cohort_summary = cohort_summary.sort_values('tenure_cohort').reset_index(drop=True)

print("\n" + "="*80)
print("COHORT RETENTION SUMMARY")
print("="*80)
print(cohort_summary[['tenure_cohort', 'total_customers', 'overall_retention_rate', 'overall_churn_rate']])

# Export for Power BI
cohort_summary.to_csv('E:\\Churn analysis\\cohort_retention_summary.csv', index=False)
print("\n✅ Exported: cohort_retention_summary.csv")

# Also keep original breakdown by contract/payment for detailed analysis
df_export = df[['tenure_cohort', 'Contract', 'PaymentMethod', 'InternetService', 'total_customers', 'retention_rate_pct']]
df_export.to_csv('E:\\Churn analysis\\cohort_detailed.csv', index=False)
print("✅ Exported: cohort_detailed.csv")
```
